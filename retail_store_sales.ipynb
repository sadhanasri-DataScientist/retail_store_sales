```python
df['Payment Method'].value_counts()
```

```python
df.groupby('Location')['Total Spent'].agg(
    Total_Revenue='sum',
    Total_Transactions='count',
    Average_Spending='mean'
)
```

```python
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])

df['Month'] = df['Transaction Date'].dt.month_name()

monthly_sales = df.groupby('Month')['Total Spent'].sum()

monthly_sales
```

```python
df['Month_Number'] = df['Transaction Date'].dt.month

monthly_sales = df.groupby(
    ['Month_Number', 'Month']
)['Total Spent'].sum().reset_index()

monthly_sales = monthly_sales.sort_values('Month_Number')

monthly_sales
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales['Month'],
    monthly_sales['Total Spent'],
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
top_products = df.groupby('Item')['Total Spent'].sum().sort_values(
    ascending=False
).head(10)

top_products
```

```python
plt.figure(figsize=(10, 6))

plt.barh(top_products.index, top_products.values)

plt.title('Top 10 Products by Revenue')
plt.xlabel('Total Revenue')
plt.ylabel('Products')

plt.gca().invert_yaxis()

plt.show()
```

```python
df['Discount Applied'].value_counts(dropna=False)
```

```python
discount_analysis = df.groupby('Discount Applied')['Total Spent'].agg(
    Total_Revenue='sum',
    Total_Transactions='count',
    Average_Spending='mean'
)

discount_analysis
```

```python
discount_analysis['Total_Revenue'].plot(
    kind='bar',
    figsize=(8, 5),
    title='Revenue: With vs Without Discount'
)

plt.xlabel('Discount Applied')
plt.ylabel('Total Revenue')
plt.xticks(rotation=0)
plt.show()
```

```python
customer_analysis = df.groupby('Customer ID')['Total Spent'].agg(
    Total_Spending='sum',
    Total_Transactions='count',
    Average_Spending='mean'
).sort_values(
    by='Total_Spending',
    ascending=False
)

customer_analysis.head(10)
```

```python
top_customers = customer_analysis.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_customers.index,
    top_customers['Total_Spending']
)

plt.title('Top 10 Customers by Total Spending')
plt.xlabel('Total Spending')
plt.ylabel('Customer ID')

plt.gca().invert_yaxis()

plt.show()
```

```python
payment_revenue = df.groupby('Payment Method')['Total Spent'].agg(
    Total_Revenue='sum',
    Average_Spending='mean'
)

payment_revenue
```

```python
payment_revenue['Total_Revenue'].plot(
    kind='bar',
    figsize=(8, 5),
    title='Revenue by Payment Method'
)

plt.xlabel('Payment Method')
plt.ylabel('Total Revenue')
plt.xticks(rotation=0)
plt.show()
```

```python
category_spending = df.groupby('Category')['Total Spent'].mean().sort_values(
    ascending=False
)

category_spending
```

```python
category_spending.plot(
    kind='bar',
    figsize=(11, 6),
    title='Average Spending by Category'
)

plt.xlabel('Category')
plt.ylabel('Average Spending')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Project Conclusion

The Retail Store Sales Analysis project was performed using Python to understand sales performance and customer purchasing behaviour.

The analysis covered the following areas:

- **Category Analysis:** Butchers recorded the highest total sales among the product categories.
- **Payment Analysis:** Cash transactions generated the highest revenue of 537,710.
- **Sales Channel Analysis:** Online sales generated higher revenue than in-store sales.
- **Monthly Sales Analysis:** January recorded the highest monthly sales of 174,421.
- **Product Analysis:** Item_25_FUR was the highest revenue-generating product with 25,256.
- **Discount Analysis:** Transactions with discounts generated slightly higher revenue than transactions without discounts.
- **Customer Analysis:** CUST_24 recorded the highest total spending of 68,452.

### Business Recommendations

2. Monitor monthly sales fluctuations to improve sales planning.
