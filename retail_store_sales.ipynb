```python
Q1 = df["Price Per Unit"].quantile(0.25)
Q3 = df["Price Per Unit"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["Price Per Unit"] < lower)|(df["Price Per Unit"] > upper)]
print("Lower boundary:", lower)
print("Upper boundary:", upper)
print("Number of outliers:", len(outliers))
```

```python
Q1 = df["Quantity"].quantile(0.25)
Q3 = df["Quantity"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["Quantity"] < lower)|(df["Quantity"] > upper)]
print("Lower boundary:", lower)
print("Upper boundary:", upper)
print("Number of outliers:", len(outliers))
```

```python
Q1 = df["Total Spent"].quantile(0.25)
Q3 = df["Total Spent"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["Total Spent"] < lower)|(df["Total Spent"] > upper)]
print("Lower boundary:", lower)
print("Upper boundary:", upper)
print("Number of outliers:", len(outliers))
```
